```python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit

qreg_q = QuantumRegister(4, 'q')
```

```python
creg_c = ClassicalRegister(4, 'c')
```

```python
circuit = QuantumCircuit(qreg_q, creg_c)
```

```python
circuit.h(qreg_q[0])
```

```python
circuit.cx(qreg_q[0], qreg_q[1])
```

```python
print(circuit.draw())
```

```python
# Measure all qubits
circuit.measure(qreg_q, creg_c)
```

```python
from qiskit_aer import AerSimulator

# Create simulator
simulator = AerSimulator()

# Run circuit 1000 times
job = simulator.run(circuit, shots=1000)

# Get result
result = job.result()

# Get counts
counts = result.get_counts(circuit)

print(counts)
```

```python
from qiskit import QuantumCircuit ,QuantumRegister,ClassicalRegister
```

```python
five_q = QuantumRegister(3,'q')
five_c = ClassicalRegister(3,'c')
circuit = QuantumCircuit(five_q,five_c)
circuit.h(five_q[0])
circuit.h(five_q[1])
circuit.ry(180,five_q[0])
circuit.cx(five_q[1],five_q[2])
circuit.x(five_q[0])
circuit.z(five_q[2])
print(circuit.draw())
```

```python
circuit.measure(five_q,five_c)
```

```python
from qiskit_aer import AerSimulator
sim = AerSimulator()
job = sim.run(circuit,shots = 1000)
result = job.result()
print(result)
counts = result.get_counts(circuit)
print(counts)
```

```python
from qiskit_ibm_runtime import QiskitRuntimeService

QiskitRuntimeService.save_account(
    token=ibm_api,
    set_as_default=True,
    overwrite=True
)
```

```python
from qiskit_ibm_runtime import QiskitRuntimeService

service = QiskitRuntimeService()
```

```python
backend = service.least_busy(
    operational=True,
    simulator=False,
    min_num_qubits=circuit.num_qubits
)

print(backend.name)
```

```python
from qiskit.transpiler import generate_preset_pass_manager

pm = generate_preset_pass_manager(
    backend=backend,
    optimization_level=1
)

isa_circuit = pm.run(circuit)

print(isa_circuit)
```

```python
from qiskit_ibm_runtime import SamplerV2 as Sampler

sampler = Sampler(mode=backend)

job = sampler.run(
    [isa_circuit],
    shots=1000
)

print("Job ID:", job.job_id())
print("Status:", job.status())
```

```python
result = job.result()

print(result)
```

```python
counts = result[0].data.c.get_counts()
print(counts)
for i in counts:
    print(i,"probality: ",counts[i])
    print(" ")
```

```python
from qiskit import QuantumRegister, ClassicalRegister,QuantumCircuit
```

```python
third_q = QuantumRegister(2,'q')
third_c = ClassicalRegister(2,'c')
circuit = QuantumCircuit(third_q,third_c)
circuit.h(third_q[0])
circuit.cx(third_q[0],third_q[1])
circuit.x(third_q[1])
print(circuit.draw())
```

```python
circuit.measure(third_q,third_c)
```

```python
from qiskit_aer import AerSimulator
sim = AerSimulator()
job = sim.run(circuit,shots = 1000)
result = job.result()

counts = result.get_counts(circuit)
print(counts)
```
