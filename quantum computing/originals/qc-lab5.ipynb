```python
from qiskit import QuantumCircuit , QuantumRegister ,ClassicalRegister
```

```python
q = QuantumRegister(5)
c=ClassicalRegister(2)
```

```python
qc = QuantumCircuit(q,c)
```

```python
qc.draw()
```

```python
#input preparing
# unknown state
# qc.x(q[1]) this makes ket 1

import numpy as np

psi = [0,1]

qc.initialize(psi, q[0])  # our own si value 

qc.barrier()

#psi = [alpha, beta]
```

```python
qc.cx(0,1)
qc.cx(0,2)

qc.barrier()

qc.x(2) # our bit error - 3rd bit

qc.barrier()

qc.cx(0,3)
qc.cx(1,3)

qc.barrier()

qc.cx(1,4)
qc.cx(2,4)
```

```python
qc.draw('mpl')
```

```python
qc.barrier()
qc.measure(3, 0)
qc.measure(4, 1)
qc.draw('mpl')
```

```python
#phase flip error

qc = QuantumCircuit(5, 2)

qc.cx(0, 1)
qc.cx(0, 2)

qc.h(0)
qc.h(1)
qc.h(2)

qc.barrier()

#qc.z(0) # our phase error bit -> 1

qc.barrier()

# Prepare ancillas |+>
qc.h(3)
qc.h(4)

# Syndrome detection
qc.cx(3, 0)
qc.cx(3, 1)

qc.cx(4, 1)
qc.cx(4, 2)

qc.barrier()

qc.h(3)
qc.h(4)


qc.draw('mpl')
```

```python
# Measure ancillas in X basis
qc.barrier()
qc.measure(3, 0)
qc.measure(4, 1)

qc.draw('mpl')
```
